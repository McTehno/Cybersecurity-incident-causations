**Analiza incidentov po državah in letih**



```python
fig, axes = plt.subplots(1, 2, figsize=(18, 6))

#top 10 držav po številu incidentov
top_countries = dataset_c_e['country'].value_counts().head(10)
sns.barplot(ax=axes[0], x=top_countries.values, y=top_countries.index, hue=top_countries.index, palette='viridis', legend=False)
axes[0].set_title('Top 10 držav po številu incidentov', fontsize=14)
axes[0].set_xlabel('Število incidentov')

#trend števila incidentov skozi leta
yearly_trends = dataset_c_e['year'].value_counts().sort_index()
sns.lineplot(ax=axes[1], x=yearly_trends.index, y=yearly_trends.values, marker='o', color='crimson', linewidth=2.5)
axes[1].set_title('Gibanje števila incidentov skozi leta (2014 - danes)', fontsize=14)
axes[1].set_xlabel('Leto')
axes[1].set_ylabel('Število incidentov')
axes[1].grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

Vidimo da največ incidentov beleži USA, sledi UK, in Rusija.

Vidimo narastek beleženja števila incidentov leta 2022, kasneje začnejo upadati

 **Različne klasifikacije incidentov**



```python
fig, axes = plt.subplots(1, 2, figsize=(18, 7))

#porazdelitev tipov incidentov
event_types = dataset_c_e['event_type'].value_counts().head(10)
sns.barplot(ax=axes[0], x=event_types.values, y=event_types.index, hue=event_types.index, palette='mako', legend=False)
axes[0].set_title('Top 10 tipov incidentov (Event Type)', fontsize=14)
axes[0].set_xlabel('Število incidentov')

#porazdelitev motivov
motives = dataset_c_e['motive'].value_counts().head(10)
sns.barplot(ax=axes[1], x=motives.values, y=motives.index, hue=motives.index, palette='rocket', legend=False)
axes[1].set_title('Top 10 motivov napadalcev (Motive)', fontsize=14)
axes[1].set_xlabel('Število incidentov')

plt.tight_layout()
plt.show()
```

**Analiza nabora podatkov EUREPOC: Akterji in pripisovanje (Attribution)**

**Padajoča matrika napadalcev in tarč (Initiator vs. Attributing/Target Country)**



```python


#filtriramo neznane vrednosti za boljšo vizualizacijo
filtered_eurepoc = dataset_eurepoc[
    (dataset_eurepoc['initiator_country'] != 'Unknown') &
    (dataset_eurepoc['attributing_country'] != 'Not available') &
    (dataset_eurepoc['attributing_country'] != 'Unknown')
]

#izberemo top 10 držav pobudnic in top 10 držav, ki izvajajo atribucijo
top_initiators = filtered_eurepoc['initiator_country'].value_counts().head(10).index
top_attributors = filtered_eurepoc['attributing_country'].value_counts().head(10).index

#pripravimo podatke za matriko
matrix_data = filtered_eurepoc[
    filtered_eurepoc['initiator_country'].isin(top_initiators) &
    filtered_eurepoc['attributing_country'].isin(top_attributors)
]

cross_tab = pd.crosstab(matrix_data['initiator_country'], matrix_data['attributing_country'])

# izris heatmapa
plt.figure(figsize=(12, 8))
sns.heatmap(cross_tab, annot=True, fmt="d", cmap="YlOrRd", cbar=True)
plt.title("Toplotni zemljevid: Država pobudnica vs. Država, ki pripisuje napad", fontsize=14)
plt.xlabel("Država, ki pripisuje (Atributing Country)", fontsize=12)
plt.ylabel("Država pobudnica (Initiator Country)", fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

USA so daleć najpogostejša drzava, ki pripisuje napade (saj imajo moćne obvešćevalne sluzbe in politiko javnega izpostavljanja napadalcev), medtem ko so drzave, kot so Kitajska, Rusija in Iran, najpogosteje oznaćene kot drzave pobudnice

Drzava pobudnica (Initiator Country):

To je drzava, iz katere napad izhaja oziroma kateri se napad pripisuje kot povzroćitelju.
Na primer, će strokovnjaki ali vlade ugotovijo, da je napad organizirala skupina, ki jo podpira Rusija ali Kitajska, bo drzava pobudnica zabelezena kot Rusija oziroma Kitajska.

Drzava, ki pripisuje napad (Attributing Country):

To je drzava (ali njena vladna agencija), ki je javno pokazala s prstom na napadalca in uradno izjavila, kdo stoji za napadom.
Pogosto je to tarća napada (npr. USA so napadene in javno obtozijo drugo drzavo) ali pa tesna zaveznica tarće (npr. UK javno podpre obtozbe svojih zaveznikov).

**Analiza tarć: Katere panoge (Industry) najbolj napadajo doloćeni akterji (Actor Type)?**


```python
# Filtriramo neznane ali splošne vrednosti za bolji prikaz
filtered_ce = dataset_c_e[
    (dataset_c_e['industry'] != 'Unknown') &
    (dataset_c_e['actor_type'] != 'Unknown') &
    (dataset_c_e['actor_type'] != 'Undetermined')
]

# Izberemo top 8 najpogostejših panog in top 5 vrst akterjev
top_industries = filtered_ce['industry'].value_counts().head(8).index
top_actors = filtered_ce['actor_type'].value_counts().head(5).index

industry_actor_data = filtered_ce[
    filtered_ce['industry'].isin(top_industries) &
    filtered_ce['actor_type'].isin(top_actors)
]

# Pripravimo pivot tabelo
industry_pivot = pd.crosstab(industry_actor_data['industry'], industry_actor_data['actor_type'])

# Izris toplotnega zemljevida
plt.figure(figsize=(12, 7))
sns.heatmap(industry_pivot, annot=True, fmt="d", cmap="Blues", cbar=True)
plt.title("Kibernetski napadi: Kdo napada katero panogo?", fontsize=14)
plt.xlabel("Vrsta napadalca (Actor Type)", fontsize=12)
plt.ylabel("Panoga tarče (Target Industry)", fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

Kibernetski kriminalci (Criminal): Najbolj intenzivno ciljajo izobraževalne ustanove (Educational Services) in finančni sektor (Finance and Insurance). Njihov glavni motiv je finančni dobiček, zato so te tarče (kjer so na voljo bodisi odkupnine bodisi neposredni finančni prenosi) najbolj privlačne.

Državno podprti akterji (State-sponsored / State-affiliated): Običajno se osredotočajo na strateške sektorje, kot so javna uprava, obramba ali kritična infrastruktura.

Haktivisti (Hacktivist): Imajo nekoliko drugačne vzorce, saj se njihovi napadi pogosto osredotočajo na javni sektor ali medije z namenom protestov oziroma političnega sporočanja.
