Distribution of attack types over time in Slovenia

```python


# Use the cleaned SI-CERT dataset
df_viz = df2.copy()

# Define periods
bins = [2014, 2017, 2020, 2024]
labels = ["2015–2017", "2018–2020", "2021–2024"]

df_viz["period"] = pd.cut(
    df_viz["report_year"],
    bins=bins,
    labels=labels
)

# Count incidents per group and period
counts = pd.crosstab(
    df_viz["period"],
    df_viz["attack_group"]
)

# Convert counts to percentages
percentages = counts.div(
    counts.sum(axis=1),
    axis=0
) * 100

# Plot
ax = percentages.plot(
    kind="bar",
    stacked=True,
    figsize=(12, 6),
    colormap="tab20"
)

plt.title("Distribution of Attack Types Over Time in Slovenia")
plt.xlabel("Reporting period")
plt.ylabel("Share of documented incidents (%)")
plt.xticks(rotation=0)
plt.legend(
    title="Attack group",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)
plt.tight_layout()
plt.show()

# Show denominators for interpretation
print(counts.sum(axis=1))
```

The graph illustrates the shifts in the relative distribution of publicly documented cyber incidents in Slovenia between 2015 and 2024. The data is divided into three time periods: 2015–2017, 2018–2020, and 2021–2024.

The results show that, during the 2015–2017 period, social engineering attacks and fraud were the most prevalent (50%), followed by malware attacks (30%) and financial attacks—specifically cryptocurrency theft (20%). In the 2018–2020 period, the distribution of attacks became more diverse; in addition to malware (30%), it included attacks targeting availability, data-focused attacks, and social engineering (each accounting for 20%), as well as unauthorized access (10%).

In the most recent period (2021–2024), the largest shares were accounted for by malware, as well as social engineering and fraud (each representing approximately 27.3%). Data-focused attacks and unauthorized access each accounted for 18.2%, while attacks targeting availability made up 9.1%.

The findings indicate that the composition of these documented incidents has evolved over time, with the proportion of malware-related attacks remaining relatively stable. Due to the small number of analyzed incidents, the findings cannot be generalized to all cyber incidents in Slovenia.

Changes in Target sectors over time in Slovenia

```python

# Create a copy
df_sector_time = df2.copy()

# Define periods
bins = [2014, 2017, 2020, 2024]
labels = ["2015–2017", "2018–2020", "2021–2024"]

df_sector_time["period"] = pd.cut(
    df_sector_time["report_year"],
    bins=bins,
    labels=labels
)

# Keep unknown target sectors visible
df_sector_time["industry"] = (
    df_sector_time["industry"].fillna("Unknown")
)

# Count incidents per sector and period
sector_counts = pd.crosstab(
    df_sector_time["period"],
    df_sector_time["industry"]
)

# Convert to percentages
sector_percentages = sector_counts.div(
    sector_counts.sum(axis=1),
    axis=0
) * 100

# Visualization
ax = sector_percentages.plot(
    kind="bar",
    stacked=True,
    figsize=(12, 6),
    colormap="tab20"
)

plt.title("Changes in Target Sectors Over Time in Slovenia")
plt.xlabel("Reporting period")
plt.ylabel("Share of incidents (%)")
plt.xticks(rotation=0)

plt.legend(
    title="Target sector",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()

print(sector_counts)
```

The graph illustrates the changes in the relative distribution of publicly documented cyber incidents across target sectors in Slovenia between 2015 and 2024. The data are divided into three time periods: 2015–2017, 2018–2020, and 2021–2024.

The results indicate that the composition of target sectors has shifted over time. In the 2015–2017 period, individuals accounted for the largest share (30%) among the identified categories, while incidents involving public administration and the education sector were absent from the analyzed sample. During the 2018–2020 period, public administration and the education sector each accounted for 20% of the documented incidents. In the most recent period (2021–2024), individuals again represented the largest share (27.3%), followed by public administration and the education sector at 18.2% each.

There is also a noticeable decrease in the proportion of incidents where the target sector is unknown—from 40% in the first two periods to approximately 18.2% in the final period.

While the results suggest shifts in the composition of publicly documented cyber incidents, the small number of cases analyzed precludes drawing reliable conclusions regarding actual trends in the threat landscape for specific sectors in Slovenia.

Police directorates

```python

# Approximate population by police directorate
population = {
    "Police Directorate Ljubljana": 670000,
    "Police Directorate Maribor": 330000,
    "Police Directorate Celje": 330000,
    "Police Directorate Kranj": 210000,
    "Police Directorate Koper": 153000,
    "Police Directorate Novo Mesto": 190000,
    "Police Directorate Nova Gorica": 118000,
    "Police Directorate Murska Sobota": 114000
}

# Count police records
directorate_counts = (
    df_period["police_directorate"]
    .value_counts()
    .reset_index()
)

directorate_counts.columns = ["directorate", "count"]

# Remove non-regional categories
directorate_counts = directorate_counts[
    ~directorate_counts["directorate"].isin([
        "General Police Directorate",
        "Not stated"
    ])
].copy()

# Match population data
directorate_counts["population"] = (
    directorate_counts["directorate"]
    .str.strip()
    .str.casefold()
    .map({
        name.casefold(): value
        for name, value in population.items()
    })
)

# Check for remaining missing values
if directorate_counts["population"].isna().any():
    print(directorate_counts[
        directorate_counts["population"].isna()
    ])
    raise ValueError("Missing population data")

# Calculate records per 100,000 inhabitants
directorate_counts["per_100k"] = (
    directorate_counts["count"]
    / directorate_counts["population"]
    * 100000
)

# Sort
directorate_counts = directorate_counts.sort_values(
    "per_100k",
    ascending=False
)

# Visualization
plt.figure(figsize=(11, 6))

ax = sns.barplot(
    data=directorate_counts,
    x="per_100k",
    y="directorate",
    color="teal"
)

for container in ax.containers:
    ax.bar_label(container, fmt="%.1f", padding=4)

plt.title(
    "Police-recorded Cybercrime per 100,000 Inhabitants "
    "(2015–2024)"
)
plt.xlabel("Records per 100,000 inhabitants")
plt.ylabel("Police directorate")
plt.tight_layout()
plt.show()

display(directorate_counts)
```

The graph shows the number of cybercrimes recorded by the police per 100,000 inhabitants, broken down by police administration in Slovenia for the period 2015–2024.

The Ljubljana Police Administration records the highest figure, with approximately 302.1 cases per 100,000 inhabitants, followed by the Nova Gorica Police Administration with 156.8 cases. The lowest figures were recorded for the Kranj Police Administration (73.8) and the Murska Sobota Police Administration (78.9).

The results indicate significant variations in the number of recorded cases relative to the population size of each area. When interpreting these findings, particular attention should be paid to the marked increase in the number of cases in 2022, which may have a substantial impact on the results for the Ljubljana Police Administration.

As the calculations are based on approximate population figures for the areas covered by each police administration, the results should be considered indicative for the time being. Before drawing final conclusions, these values ​​should be verified against official demographic data.

Police coding labels by offence year

```python
years = [2021, 2022, 2023]

methods_3y = police_methods.loc[police_methods["offence_year"].isin(years)]
top_8 = methods_3y["method"].value_counts().head(8).index

method_counts = pd.crosstab(
    methods_3y["method"],
    methods_3y["offence_year"]
).reindex(index=top_8, columns=years, fill_value=0)

records_per_year = (
    df_period.groupby("offence_year")["police_record_id"]
    .nunique()
    .reindex(years)
)

method_pct = method_counts.div(records_per_year, axis="columns") * 100
method_pct.columns = [f"{year}\n(n={records_per_year[year]:,})" for year in years]

plt.figure(figsize=(9, 5))
sns.heatmap(
    method_pct, annot=True, fmt=".1f", cmap="Blues",
    vmin=0, vmax=100, linewidths=0.5,
    cbar_kws={"label": "Share of yearly records (%)"}
)
plt.title("Police coding labels by offence year")
plt.xlabel("Offence year and number of records")
plt.ylabel("")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

The heat map displays the proportions of specific police classifications for cybercrimes in Slovenia during the 2021–2023 period.

The results reveal a marked anomaly in 2022, when the classification "Data input" accounted for as much as 84.1% of all 1,330 records, whereas its share was only 2.8% in 2021 and 5.1% in 2023. Concurrently, the relative share of most other classifications decreased significantly in 2022. For instance, the classification "Unauthorized access" accounted for 50.0% of records in 2021, 7.6% in 2022, and 43.2% in 2023.

This distribution suggests that the sharp increase in the number of records in 2022 was not evenly spread across the various crime classifications but was, instead, primarily linked to the "Data input" category.

A comparison with other visualizations reveals that the approximate number of records labeled "Data input" in 2022 aligns with the number of records for offenses committed in 2022 that have a final document dated 2024. However, further analysis of individual records is required to confirm this connection. The results therefore underscore the importance of verifying the structure and method of recording police data before interpreting a marked increase in the number of records as an actual rise in cybercrime in Slovenia.
